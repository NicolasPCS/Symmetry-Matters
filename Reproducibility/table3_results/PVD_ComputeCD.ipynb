```python
import os
import sys
import statistics
from pathlib import Path

import numpy as np

PROJECT_ROOT = Path.cwd()
if PROJECT_ROOT.name == "Table3_Results":
    PROJECT_ROOT = PROJECT_ROOT.parent.parent
elif PROJECT_ROOT.name == "Reproducibility":
    PROJECT_ROOT = PROJECT_ROOT.parent

SYMMETRY_PROTOCOL_DIR = PROJECT_ROOT / "Symmetry_Measurement_Protocol"
if str(SYMMETRY_PROTOCOL_DIR) not in sys.path:
    sys.path.append(str(SYMMETRY_PROTOCOL_DIR))

from ChamferDistance import chamfer_distance
```

# Original dataset

## Airplane

```python
one_airplane_input_original_objects = "/home/ncaytuir/data/Datasets/Resultados_PVD/NewExp_ArchitectureAssessing/original_ckpts/airplane/original"
one_airplane_input_mirrored_objects = "/home/ncaytuir/data/Datasets/Resultados_PVD/NewExp_ArchitectureAssessing/original_ckpts/airplane/mirrored"

one_airplane_files_original_objects = sorted([f for f in os.listdir(one_airplane_input_original_objects) if f.endswith(".npy")])
one_airplane_files_mirrored_objects = sorted([f for f in os.listdir(one_airplane_input_mirrored_objects) if f.endswith(".npy")])

one_cd_results_airplane = []

for i in one_airplane_files_original_objects:
    file_path_pc_1 = os.path.join(one_airplane_input_original_objects, i)
    file_path_pc_2 = os.path.join(one_airplane_input_mirrored_objects, i)

    #print(file_path_pc_1)
    #print(file_path_pc_2)
    
    original_pc = np.load(file_path_pc_1)
    mirrored_pc = np.load(file_path_pc_2)

    if original_pc.shape[0] == 15000:
        pass

    # Compute CD
    cd_result = chamfer_distance(original_pc, mirrored_pc)

    one_cd_results_airplane.append(cd_result)

print("\nAverage mean " + str(len(one_airplane_files_original_objects)) +":", statistics.mean(one_cd_results_airplane))
print("Average mean/max:", max(one_cd_results_airplane))
print("Average mean/min:", min(one_cd_results_airplane))
```
